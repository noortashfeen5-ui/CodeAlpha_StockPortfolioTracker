```python
stock_prices = {
    "AAPL": 180,
    "TSLA": 250,
    "GOOGL": 140,
    "AMZN": 180,
    "MSFT": 420
}

print("Available Stocks:")
print(stock_prices)
```

```python
stock = input("Enter the stock name: ").upper()

print("You selected:", stock)
```

```python
quantity = int(input("Enter the quantity: "))

print("Quantity:", quantity)
```

```python
if stock in stock_prices:
    total = stock_prices[stock] * quantity
    print("Total Investment: $", total)
else:
    print("Stock not found.")
```

```python
another = input("Do you want to add another stock? (yes/no): ").lower()

if another == "yes":
    stock2 = input("Enter the second stock name: ").upper()
    quantity2 = int(input("Enter the quantity: "))

    if stock2 in stock_prices:
        total2 = stock_prices[stock2] * quantity2
        print("Investment for", stock2, "is: $", total2)
    else:
        print("Stock not found.")
else:
    print("Thank you for using Stock Portfolio Tracker!")
```

```python
portfolio_total = total + total2

print("Total Portfolio Investment: $", portfolio_total)
```

```python
file = open("portfolio.txt", "w")

file.write("Stock Portfolio Tracker\n")
file.write("-----------------------\n")
file.write(f"{stock}: {quantity} shares = ${total}\n")
file.write(f"{stock2}: {quantity2} shares = ${total2}\n")
file.write(f"Total Investment = ${portfolio_total}\n")

file.close()

print("Portfolio saved successfully!")
```

```python
stock_prices = {
    "AAPL": 180,
    "TSLA": 250,
    "GOOGL": 140,
    "AMZN": 180,
    "MSFT": 420
}

print("Available Stocks:")
print(stock_prices)

stock = input("Enter the stock name: ").upper()
quantity = int(input("Enter the quantity: "))

if stock in stock_prices:
    total = stock_prices[stock] * quantity
    print("Investment for", stock, "is: $", total)

    another = input("Do you want to add another stock? (yes/no): ").lower()

    if another == "yes":
        stock2 = input("Enter the second stock name: ").upper()
        quantity2 = int(input("Enter the quantity: "))

        if stock2 in stock_prices:
            total2 = stock_prices[stock2] * quantity2
            portfolio_total = total + total2

            print("Investment for", stock2, "is: $", total2)
            print("Total Portfolio Investment: $", portfolio_total)

            with open("portfolio.txt", "w") as file:
                file.write("Stock Portfolio Tracker\n")
                file.write("-----------------------\n")
                file.write(f"{stock}: {quantity} shares = ${total}\n")
                file.write(f"{stock2}: {quantity2} shares = ${total2}\n")
                file.write(f"Total Investment = ${portfolio_total}\n")

            print("Portfolio saved successfully!")

        else:
            print("Stock not found.")
    else:
        print("Thank you for using Stock Portfolio Tracker!")

else:
    print("Stock not found. Please enter a valid stock name.")
```
